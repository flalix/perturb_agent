```python
from inmoose.limma import lmFit, eBayes, topTable, makeContrasts, contrasts_fit

# def calc_limma_inmoose(ncluster:int) -> pd.DataFrame:

ncluster = 1

tumor_samples = gdc.df_hca[gdc.df_hca.cluster == ncluster]['sample']

if len(tumor_samples) == 0:
    print(f"Error: {ncluster} has no tumor cols")
else:
    print(f"Error: {ncluster} has {len(tumor_samples)} tumor cols")

sel_cols = list(tumor_samples) + list(normal_samples)

df_combat_nclu = gdc.df_combat[sel_cols].copy()
df_combat_nclu.index = gdc.df_combat.geneid

metadata = pd.DataFrame(
    {
        "group": (
            ["Tumor"] * len(tumor_samples) +
            ["Normal"] * len(normal_samples)
        )
    },
    index=list(tumor_samples) + list(normal_samples),
)

metadata = metadata.loc[df_combat_nclu.columns]

# Intercept + binary tumor coefficient
# Normal is the reference condition
design = pd.DataFrame(
    {
        "Intercept": 1.0,
        "Tumor_vs_Normal": (
            metadata["group"].eq("Tumor").astype(float)
        ),
    },
    index=metadata.index,
)

assert df_combat_nclu.columns.equals(design.index)

fit = lmFit(df_combat_nclu, design)

# InMoose supports trend=True for logCPM-like data and calculates moderated statistics through empirical-Bayes variance shrinkage.
coefficient_names = ["Intercept", "Tumor_vs_Normal"]

fit.coefficients.columns = coefficient_names
fit.stdev_unscaled.columns = coefficient_names

fit_bayes = eBayes(
    fit,
    trend=True,
    robust=False,
)

results = topTable(
    fit_bayes,
    coef="Tumor_vs_Normal",
    number=df_combat_nclu.shape[0],
    # genelist=gene_annotations,
    adjust_method="fdr_bh",
    sort_by="P",
)

df_lfc_ori = pd.DataFrame(
    data=results.to_numpy(copy=True),
    index=results.index.copy(),
    columns=results.columns.copy(),
)

df_lfc_ori.head(3)
```
